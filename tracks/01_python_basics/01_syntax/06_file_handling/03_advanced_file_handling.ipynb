# Binary Mode

```python
value = 543210000000
bytes_value = value.to_bytes(length=5, byteorder='big')
"""
Length of bytes object to use. An OverflowError is raised if the
integer is not representable with the given number of bytes.  Default
is length 1.
"""
with open('example.bin', 'wb') as file:
    file.write(bytes_value)
```

```python
with open('example.bin', 'rb') as file:
    content = file.read()
    value = int.from_bytes(content, byteorder='big')
print(value)
```

```python
with open('example.txt','w') as file:
    value_str = str(value)
    file.write(value_str)
```

# Navigation

```python
with open('example2.txt', 'w') as file:
    file.write('Line 1\nLine 2\nLine 3')

with open('example2.txt', 'r') as file:
    print(file.tell())
    print(file.readline().strip())
    print(file.tell())
    print(file.readline().strip())
    print(file.tell())
    print(file.readline().strip())
    print(file.tell())

```

```python
# Using various file methods
with open('example2.txt', 'r') as file:
    print(file.read())  # Read entire file
    print(file.tell())  # 22
    file.seek(2)  # Move file pointer to position 2
    print(file.readline())  # Read from position 2
    file.seek(0)
    print(file.readlines())  # Read all lines into oryg_list list
```
