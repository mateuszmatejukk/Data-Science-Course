## 1. Working with Paths


### Exercise 1
Write a program to print the absolute path of the current working directory.


```
import os

print(os.path.abspath(os.getcwd()))
```

### Exercise 2
Create a new directory named `test_directory` in the current working directory. Check if it exists, and print a message confirming its creation.


```
import os

directory_name = "test_directory"
directory_path = os.path.join(os.getcwd(), directory_name)

os.makedirs(directory_path, exist_ok=True)

if os.path.exists(directory_path) and os.path.isdir(directory_path):
    print(f"Directory '{directory_name}' was created successfully at: {directory_path}")
else:
    print(f"Failed to create directory '{directory_name}'.")
```

### Exercise 3
Write a program to create a file named `test_file.txt` in the `test_directory`, write "Hello, World!" into it, and close the file.


```
import os

directory_path = os.path.join(os.getcwd(), "test_directory")
file_path = os.path.join(directory_path, "test_file.txt")

with open(file_path, "w", encoding="utf-8") as file:
    file.write("Hello, World!")

print("File created and text written successfully.")
```

### Exercise 4
Change the permissions of the `test_file.txt` to make it read-only. Verify the permissions programmatically.


```
import os
import stat

file_path = os.path.join(os.getcwd(), "test_directory", "test_file.txt")

os.chmod(file_path, stat.S_IREAD)

file_stats = os.stat(file_path)
is_read_only = not (file_stats.st_mode & stat.S_IWRITE)

if is_read_only:
    print("The file is read-only.")
else:
    print("The file is not read-only.")
```

## 2. File Operations


### Exercise 5
Write a program to read and print the content of the file `test_file.txt`.


```
import os

file_path = os.path.join(os.getcwd(), "test_directory", "test_file.txt")

with open(file_path, "r", encoding="utf-8") as file:
    content = file.read()

print(content)
```

### Exercise 6
Write a program to append "This is a new line." to `test_file.txt` and print the updated content.


```
import os
import stat

file_path = os.path.join(os.getcwd(), "test_directory", "test_file.txt")

# przywrócenie zapisu, jeśli plik był read-only
os.chmod(file_path, stat.S_IWRITE)

with open(file_path, "a", encoding="utf-8") as file:
    file.write("\nThis is a new line.")

with open(file_path, "r", encoding="utf-8") as file:
    content = file.read()

print(content)
```

### Exercise 7
Write a program to read the content of `test_file.txt` into a list, where each element of the list is a line from the file.


```
with open("test_file.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

print(lines)
```

### Exercise 8
Write a program to count the number of occurrences of the word "Hello" in `test_file.txt`.


```
count = 0

with open("test_file.txt", "r", encoding="utf-8") as file:
    for line in file:
        count += line.count("Hello")

print("Number of occurrences of 'Hello':", count)
```

### Exercise 9
Write a program to delete `test_file.txt` and confirm its deletion by checking its existence.


```
import os

filename = "test_file.txt"

if os.path.exists(filename):
    os.remove(filename)
    print(f"{filename} has been deleted.")
else:
    print(f"{filename} does not exist.")

if not os.path.exists(filename):
    print("Deletion confirmed.")
else:
    print("File still exists.")
```
