# Removing Files

```python
import os
absolute_path = os.path.abspath('dir\\file.dupa')
if os.path.exists(absolute_path):
    os.remove(absolute_path)
    print(f"file removed")
else:
    print("file does not exists")
```

# Reading files

```python
absolute_path = os.path.abspath('dir\\file.txt')
file = open(absolute_path, 'w')
file.write('line1\n')
file.write('line2\n')
file.write('line3\n')
file.close()
```

```python
file = open(absolute_path, 'r')
content = file.read()
file.close()
```

```python
file = open(absolute_path, 'r')
lines = file.readlines()
lines
```

```python
with open(absolute_path, 'r') as file:
    content = file.read()
print(content)
file.read() # you cannot read closed file
```
