## File Permissions

In Linux, file permissions are represented using a **9-character format** and categorized into three groups:

### Permission Format
**Format**: `rwxrwxrwx`
- **Owner**: `rwx` (Read, Write, Execute)
- **Group**: `rwx` (Read, Write, Execute)
- **Others**: `rwx` (Read, Write, Execute)

### Binary Representation
Each permission can be expressed in binary:
- **Read (r)** = 4 (binary `100`)
- **Write (w)** = 2 (binary `010`)
- **Execute (x)** = 1 (binary `001`)

Combined permissions are calculated by adding the binary values:
- **Read + Write + Execute** = 7 (binary `111`)
- **Read + Execute** = 5 (binary `101`)
- etc

### User Categories
1. **Owner** - The user who created the file with full control.
2. **Group** - A set of users with shared permissions for the file.
3. **Others** - All other users with limited access.

### Example
For a permission setting of `754`:
- Owner: `rwx` (7) - Full control
- Group: `r-x` (5) - Read and Execute
- Others: `r--` (4) - Read only
