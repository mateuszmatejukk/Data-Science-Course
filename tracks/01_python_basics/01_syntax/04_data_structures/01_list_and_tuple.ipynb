# List vs Tuple

Similarities:
- ordered collections of items
- can store non-homogeneous data types
- can store non-unique values
- can be nested
- can be indexed and sliced
- can be iterated over
- can be unpacked

Difference:
- list is mutable, tuple is immutable


```python
l = [1, 2, 3]
t = (1, 2, 3)
type(l), type(t)
```

```python
l, t
```

```python
tuple(l)
```

```python
list(t)
```

# Differences between list and tuple

```python
# list i mutable - meaning it can be changed after assignment
l = [1,2,5]
l[2] = 3
l
```

```python
# tuple i unmutable - meaning it can't be changed after assignment
t = (1,2,5)
t[2] = 3 # TypeError: 'tuple' object does not support item assignment
t
```
