# Set operation

```python
s1, s2 = {1,2,3}, {3,4,5}
s1.union(s2)
```

```python
s1.intersection(s2)
```

```python
s1.difference(s2)
```

```python
s2.difference(s1)
```

```python
s1.symmetric_difference(s2)
```

```python
s1.issubset({1, 2, 3, 4})
```

```python
s1.issuperset({1,2})
```

```python
s1.isdisjoint({4,5})
```
