### Exercise 9

`{'a': 1, 'b': 2}` and `{'c': 3, 'd': 4}`.


````python
d1 = {'a': 1, 'b': 2}
d2 = {'c': 3, 'd': 4}
d3 = {}
for k in d1:
    d3.update(d1)
for k in d2:
    d3.update(d2)
d3
````

Write a program to calculate the sum and product of all items in the dictionary:

### Exercise 15
Write a program to generate a dictionary with numbers as keys and their squares as values.
Example: For `n=5`, the output should be `{1: 1, 2: 4, 3: 9, 4: 16, 5: 25}`.


````python
d = {}
num = int(input('Enter a number: '))
for num in range(1,num + 1):
    d[num] = num ** 2
d
````

### Exercise 16
Write a program to check if a given string contains only uppercase letters, only lowercase letters, or both.


````python
d = {}
txt = input('Enter a text: ')
for l in txt:
    if l.islower():
        litera = 'lower'
    if l.isupper():
        litera = 'upper'
    d.update({l:litera})
d
````

### Exercise 17
Duplicate a dictionary, but only include keys with values greater than 10.
Example:
- Input: `{'a': 5, 'b': 15, 'c': 20}`
- Output: `{'b': 15, 'c': 20}`


````python
d1 = {'a': 5, 'b': 15, 'c': 20}
d2 = {}
for k,v in d1.items():
    if v > 10:
        d2.update({k:v})
d2
````
