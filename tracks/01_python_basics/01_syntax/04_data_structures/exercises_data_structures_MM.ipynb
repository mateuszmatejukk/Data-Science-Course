````
# Data Structures Exercises
````

## 1. Lists and Tuples


### Exercise 1
Create a list containing the first 10 square numbers and print the result.


### Exercise 2
Write a program to find the maximum and minimum elements in a list.


### Exercise 3
Convert the tuple `(1, 2, 3, 4, 5)` into a list, append the value `6`, and convert it back to a tuple.


### Exercise 4
Write a program to print all books with more than 200 pages.
```
books = [
    {"title": "Book A", "pages": 50},
    {"title": "Book B", "pages": 300},
    {"title": "Book C", "pages": 250}
]
```


## 2. Sets


### Exercise 5
Remove duplicate numbers from the list `[1, 2, 2, 3, 4, 4, 5]` using a set.


### Exercise 6
Write a program to find the union, intersection, and difference of two sets: `{1, 2, 3}` and `{3, 4, 5}`.


## 3. Dictionaries


### Exercise 7
Create a dictionary where the keys are numbers from 1 to 5 and the values are their squares.


### Exercise 8
Sort the dictionary `{'b': 2, 'a': 1, 'c': 3}` by its values.


### Exercise 9
Write a program to concatenate two dictionaries:
`{'a': 1, 'b': 2}` and `{'c': 3, 'd': 4}`.


### Exercise 10
Create a deep copy of a dictionary and modify the original to verify the deep copy remains unchanged.


### Exercise 11
Write a program to calculate the sum and product of all items in the dictionary:
`{'a': 1, 'b': 2, 'c': 3}`.


## 4. Comprehensions


### Exercise 12
Create a list of cubes of numbers from 1 to 10 using list comprehension.


### Exercise 13
Create a dictionary where the keys are characters in the string "python" and the values are their ASCII codes.


## 5. Practical Applications


### Exercise 14
Write a program to filter out books with fewer than 100 pages.
```
books = [
    {"title": "Book A", "pages": 50},
    {"title": "Book B", "pages": 300},
    {"title": "Book C", "pages": 150}
]
```


### Exercise 15
Write a program to generate a dictionary with numbers as keys and their squares as values.
Example: For `n=5`, the output should be `{1: 1, 2: 4, 3: 9, 4: 16, 5: 25}`.


### Exercise 16
Write a program to check if a given string contains only uppercase letters, only lowercase letters, or both.


### Exercise 17
Duplicate a dictionary, but only include keys with values greater than 10.
Example:
- Input: `{'a': 5, 'b': 15, 'c': 20}`
- Output: `{'b': 15, 'c': 20}`
