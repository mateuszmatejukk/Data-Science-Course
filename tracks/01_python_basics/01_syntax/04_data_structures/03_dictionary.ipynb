# Iterating

```python
d = dict(
    key1='val1',
    key2='val2',
    key3='val3',
)
d
```

```python
for k in d:
    print(k, d[k])
```

```python
for item in d.items():
    print(item[0], item[1])
```

```python
for item in d.items():
    k = item[0]
    v = item[1]
    print(k, v)
```

```python
for item in d.items():
    k, v = item[0], item[1]
    print(k, v)
```

```python
for item in d.items():
    k, v = item # unpacking
    print(k, v)
```

Recommended approach to display all key, value pairs

```python
for k, v in d.items():
    print(k, v)
```

values only

```python
for k, v in d.items():
    print(v)
```

```python
for _, v in d.items():
    print(v)
```

```python
for v in d.values():
    print(v)
```

```python
for k, v in d.items():
    print(k, v)
```

```python
cnt = 7
for v in 'abc':
    print(cnt, v)
    cnt +=1
```

Recommended approach to enumerate over values (list values with index)

```python
for i, v in enumerate('abc', start=7):
    print(i, v)
```
