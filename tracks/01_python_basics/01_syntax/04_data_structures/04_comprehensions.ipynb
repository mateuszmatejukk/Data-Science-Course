# List comprehension

```python
l =[]
for i in range(5):
    l.append(i**2)
l # 0, 1, 2, 4, 9, 16
```

```python
l = [i**2 for i in range(5)]
l
```

# Set comprehension

```python
s = set()
for i in range(15):
    if i%3==0:
        s.add(i)
s # 1, 2, 4, 5, 7, 8
```

```python
s = set(number for number in range(15) if number%3==0)
# create SET of NUMBER FOR each NUMBER IN RANGE 0-15 IF that NUMBER %3 IS EQUAL to 0
s
```

# Dict comprehension

```python
d = {}
for i in range(5):
    d[i] = i**2
d
```
