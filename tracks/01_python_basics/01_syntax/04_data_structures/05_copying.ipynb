# Copying

## simple data types (e.g. integer)

```python
a = 1
b = a  # copy value
print(a, b)
```

```python
a += 1
print(a, b)
```

## collections (e.g. lists, dicts) store references

```python
l = [1, 2, 3]
k = l  # tworzy referencję
print(l, k)
```

```python
l.append(4)
print(l, k)
```

# Using copy to create a separate object

```python
l = [1, 2, 3]
m = l  # reference
n = l.copy()  # create a shallow copy
l.append(4)
print(l)  # [1, 2, 3, 4]
print(m)  # [1, 2, 3, 4]
print(n)  # [1, 2, 3]
```

```python
print(id(l), id(m), id(n))
```

```python
id(l) == id(m)
```

```python
id(l) == id(n)
```

# Nested list

```python
l = [1, 2, ["a", "b"]]
m = l.copy()  # shallow copy
l.append(4)
print(l)
print(m)
```

```python
l = [1, 2, ["a", "b"]]
m = l.copy()  # shallow copy
l[2].append("c")
print(l)  # [1, 2, ['a', 'b', 'c']]
print(m)  # [1, 2, ['a', 'b']] < wrong because of shallow copy
```

```python
import copy

l = [1, 2, ["a", "b"]]
m = copy.deepcopy(l)  # deep copy
l[2].append("c")
print(l)  # [1, 2, ['a', 'b', 'c']]
print(m)  # [1, 2, ['a', 'b']] < correct because of deep copy
```

# Summary

```python
import copy

l = [1, 2, ["a", "b"]]
reference = l
shallow_copy = copy.copy(l)  # same as l.copy()
deep_copy = copy.deepcopy(l)
# l.append(3)
l[2].append("c")
print(f"{l=}")  # Output: [1, 2, ['a', 'b', 'c']]
print(f"{reference=}")  # Output: [1, 2, ['a', 'b', 'c']]
print(f"{shallow_copy=}")  # Output: [1, 2, ['a', 'b', 'c']]
print(f"{deep_copy=}")  # Output: [1, 2, ['a', 'b']]

```
