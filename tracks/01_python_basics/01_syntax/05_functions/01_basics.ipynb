# Arbitrary arguments - *args

```python
def pizza(pizza_name, *toppings):
    print(f"Making {pizza_name}")
    for topping in toppings:
        print(f"- {topping}")

pizza("capriciosa", "pepperoni", "cheese", "mushroms")
```

```python
pizza("capriciosa", "pepperoni", "cheese")
```

```python
pizza("Empty Pizza")
```

# Keyword Arguments - **kwargs

```python
def pizza(pizza_name, *toppings, **settings):
    print(f"Making {pizza_name}")
    for topping in toppings:
        print(f"- {topping}")
    print(f"Bake at {settings.get('temp', 180)} degrees")
    print(f"Bake for {settings.get('time', 3)} minutes")

pizza("capriciosa", "pepperoni", "cheese", "mushrooms", temp=200)
```

```python
pizza("capriciosa", "pepperoni", temp=200, time=20)
```

# Mutable Default Argument

```python
def append_to(l):
    l.append(9)
    return l

print(append_to([1,2])) # [1,2,9]
print(append_to([1,2,3])) # [1,2,3,9]
```

```python
# Dangerous
def append_to(l=[]): # using mutable default argument is 99% wrong
    l.append(9)
    return l

print(append_to([1,2])) # [1,2,9]
print(append_to()) # [9]
print(append_to()) # [9, 9]
print(append_to()) # [9, 9, 9]
```

```python
# Correct
def append_to(l=None):
    if l is None:
        l = []
    l.append(9)
    return l

print(append_to([1,2])) # [1,2,9]
print(append_to()) # [9]
print(append_to()) # [9]
print(append_to()) # [9]
```
