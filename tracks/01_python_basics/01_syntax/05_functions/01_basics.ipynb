# defining and calling a function

```python
def welcome():
    print("Hello, World!")


welcome()
welcome()
```

# argments (aka parameters)

```python
def fun(name):
    print("Hello, " + name + "!")


# fun() # zwróci błąd: missing 1 required positional argument: 'name'
fun("Mati")  # równoznaczne z tym co niej
fun(name="Mati")  # w 90% przypadków polecam to
```

# default values

```python
def fun(name="World"):
    print("Hello, " + name + "!")


fun()
fun("Mati")
fun(name="Mati")
```

# more Arguments

```python
def card(surname, name):
    print(f"{name.title()} {surname.title()}")


card("jan", "kowalski")  # argumenty pozycyjne
card(name="jan", surname="kowalski")  # argumenty nazwane <-- recommended
```
