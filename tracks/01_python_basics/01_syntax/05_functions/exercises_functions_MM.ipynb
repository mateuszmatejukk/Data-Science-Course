## 1. Basics of Functions


### Exercise 1
Write a function to find the maximum of three numbers.


```python
def biggest_num(a,b,c):
    return max(a,b,c)
biggest_num(1,2,3)
```

### Exercise 2
Write a function to calculate the sum of all numbers in a list.


```python
def sum_list(numbers):
    suma = 0
    for n in numbers:
        suma += n
    return suma

sum_list([1,2,3,4])
```

### Exercise 3
Write a function to calculate the product of all numbers in a list.


```python
def multiply_list(numbers):
    calculation = 1
    for n in numbers:
        calculation *= n
    return calculation


multiply_list([1, 2, 3, 4, 5])
```
