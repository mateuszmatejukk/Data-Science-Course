## 1. Basics of Functions


### Exercise 1
Write a function to find the maximum of three numbers.


```python
def nums(first,second,third):
    biggest = max(first,second,third)
    return biggest

nums(1,2,3)
```

### Exercise 2
Write a function to calculate the sum of all numbers in a list.


```python
def sum_of_list(l):
    sum = 0
    for nums in l:
        sum += nums
    return sum

sum_of_list([1,2,3,4,5,6])
```

### Exercise 3
Write a function to calculate the product of all numbers in a list.


```python
def product_of_list(l):
    product = 1
    for nums in l:
        product *= nums
    return product

product_of_list([1,2,3,4,5,6])
```

### Exercise 4
Write a recursive function to reverse a string.


```python
def reversed(text):
    if len(text) == 0:
        return text
    else:
        return text[-1] + reversed(text[:-1])

reversed('kajak')
```

### Exercise 5
Write a recursive function to calculate the factorial of a given number.


```python
def factorial(num):
    if num == 0:
        return 1
    else:
        return num * factorial(num - 1)

factorial(5)
```

## 2. Lambda Functions


### Exercise 8
Use a lambda function to filter out even numbers from a given list.


```python
def even(l):
    l2 = []
    for nums in l:
        if nums % 2 == 0:
            l2.append(nums)
    return(l2)

even([1,2,3,4,5,6])
```

```python
lambda_even = lambda l: [nums for nums in l if nums % 2 == 0]
lambda_even([1,2,3,4,5,6])
```

### Exercise 9
Write a lambda function to count the number of uppercase and lowercase letters in a string.


```python
def counter(text):
    upper_counter = 0
    lower_counter = 0
    for elements in text:
        if elements.islower():
            lower_counter += 1
        elif elements.isupper():
            upper_counter += 1

    return lower_counter, upper_counter

counter('DJIfnvhgfnuafASd')
```

## 3. Functional Programming Tools


### Exercise 12
Write a function using `map` to square all numbers in a list.


```python
def container(num):
    return num ** 2
list(map(container, [1,2,3,4,5,6]))
```

### Exercise 13
Write a function using `filter` to extract only even numbers from a list.


```python
def extracter(num):
    if num % 2 == 0:
        return True
    else:
        return False
list(filter(extracter,[1,2,3,4,5,6]))
```

### Exercise 14
Write a function using `reduce` to find the sum of all numbers in a list.


```python
from functools import reduce

def add(a,b):
    print(f'{a} + {b} = {a+b}')
    return a+b

reduce(add, [1,2,3,4,5,6,7])
```

### Exercise 15
Combine `map`, `filter`, and `reduce` to process a list:
- Square each number.
- Filter numbers greater than 10.
- Compute their sum.


```python
def first(num):
    return num ** 2
square = list(map(first,[1,2,3,4,5]))
def filtered_list(num):
    if num < 10:
        return False
    else:
        return True

new_list = list(filter(filtered_list,square))
print(new_list)

from functools import reduce
def sum_of_both_nums(a,b):
    print(f"{a} + {b} = {a + b}")
    return a + b

reduce(sum_of_both_nums,new_list)
```
