# Namespace and Scope in Python: LEGB Rule
- Python resolves names using four nested scopes:
  - **L**: Local — inside the current function.
  - **E**: Enclosing — inside any outer functions.
  - **G**: Global — top-level of the module/file.
  - **B**: Built-in — names like `len`, `print`, `sum`.

```python
def fun():
    x = 'local'
    return x

fun()
```

```python
def outer():
    x = 'enclosing'
    def inner():
        return x

    return inner()

outer()
```

```python
x = 'global'

def show():
    def snow2():
        return x
    return snow2()
show()
```

```python
# print definition is in a different place, but print is builtin
print('builtin')
```
