```python
students = [
    {
        'name':'John',
        'grade':70,
    },
    {
        'name':'Alice',
        'grade':90,
    },
    {
        'name':'Bob',
        'grade':85,
    }
]


def sorting_fun(student_dict):
    return student_dict['name']


sorted(students, key=sorting_fun,reverse=True)
```

```python

sorted(students, key=lambda student_dict: student_dict['name'] ,reverse=True)
```
