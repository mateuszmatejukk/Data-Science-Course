# map

```python
l = range(8)
fun = lambda x: (x-1)/2
l_mapped = []
for i in l:
    l_mapped.append(fun(i))
l_mapped
```

```python
l_mapped = map(fun, l)
list(l_mapped)
```

# filter

```python
filter_fun = lambda x: x%2==1
list(filter(filter_fun, l))
```

# reduce

```python
from functools import reduce

def add(a, b):
    print(f'{a} + {b} = {a+b}')
    return a+b


reduce(add, l)
# 0,1,2,3,4,5,6,7
# 0,1 --> 0+1 = *1
# *1,2 --> 1+2 = 3
# 3, 3 --> 3+3 = 6
# 6,4 --? 6+4 = 10
# ... 28
```

```python
reduce(lambda a,b:a+b, l)
```
