# map

```python
l = range(8)
fun = lambda x: (x-1)/2
l_mapped = []
for i in l:
    l_mapped.append(fun(i))
l_mapped
```

# reduce

```python
from functools import reduce

def add(a, b):
    print(f'{a} + {b} = {a+b}')
    return a+b


reduce(add, l)
# 0,1,2,3,4,5,6,7
# 0,1 --> 0+1 = *1
# *1,2 --> 1+2 = 3
# 3, 3 --> 3+3 = 6
# 6,4 --? 6+4 = 10
# ... 28
```

# Combining Funtional Tools

```python
l = range(8)
    
def fun(x):
    return (x-1)/2
def filter_fun(x): return x%2==1

# x: 0,1,2,3,4,5,6,7
# fun(x): -0.5,0,0.5,1,1.5,2,2.5,3
# filter_fun(fun(x)): 1, 3
# sum: 4


total_sum = 0
for x in l:
    if filter_fun(result := fun(x)):
        total_sum += result
total_sum
```

```python
sum([result for x in l if filter_fun(result := fun(x))])
```

```python
from functools import reduce

sum_fun = lambda x, y: x+y
reduce(sum_fun, filter(filter_fun, map(fun, l)))
```

```python
list(map(fun, filter(filter_fun, l)))
```
