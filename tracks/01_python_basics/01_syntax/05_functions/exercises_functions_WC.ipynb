## 1. Basics of Functions


### Exercise 1
Write a function to find the maximum of three numbers.


```
def nums(first, second, third):
    biggest = max(first, second, third)
    return biggest


nums(1, 2, 3)
```

### Exercise 2
Write a function to calculate the sum of all numbers in a list.


```
def sum_of_list(l):
    sum = 0
    for nums in l:
        sum += nums
    return sum


sum_of_list([1, 2, 3, 4, 5, 6])
```

### Exercise 3
Write a function to calculate the product of all numbers in a list.


```
def product_of_list(l):
    product = 1
    for nums in l:
        product *= nums
    return product


product_of_list([1, 2, 3, 4, 5, 6])
```

### Exercise 4
Write a recursive function to reverse a string.


```
def reversed(text):
    if len(text) == 0:
        return text
    else:
        return text[-1] + reversed(text[:-1])


reversed('kajak')
```

## 2. Lambda Functions


### Exercise 8
Use a lambda function to filter out even numbers from a given list.


```
def even(l):
    l2 = []
    for nums in l:
        if nums % 2 == 0:
            l2.append(nums)
    return (l2)


even([1, 2, 3, 4, 5, 6])
```

### Exercise 9
Write a lambda function to count the number of uppercase and lowercase letters in a string.


```
def counter(text):
    upper_counter = 0
    lower_counter = 0
    for elements in text:
        if elements.islower():
            lower_counter += 1
        elif elements.isupper():
            upper_counter += 1

    return lower_counter, upper_counter


counter('DJIfnvhgfnuafASd')
```
