# Functions and Functional Programming Exercises


## 1. Basics of Functions


### Exercise 1
Write a function to find the maximum of three numbers.


### Exercise 2
Write a function to calculate the sum of all numbers in a list.


### Exercise 3
Write a function to calculate the product of all numbers in a list.


### Exercise 4
Write a recursive function to reverse a string.


### Exercise 5
Write a recursive function to calculate the factorial of a given number.


### Exercise 6
Write a function to check whether a number is in a given range.


### Exercise 7
Write a function to remove a specific element from a list by value.


## 2. Lambda Functions


### Exercise 8
Use a lambda function to filter out even numbers from a given list.


### Exercise 9
Write a lambda function to count the number of uppercase and lowercase letters in a string.


### Exercise 10
Write a lambda function to extract unique elements from a list.


### Exercise 11
Use a lambda function to validate if a number is prime.


## 3. Functional Programming Tools


### Exercise 12
Write a function using `map` to square all numbers in a list.


### Exercise 13
Write a function using `filter` to extract only even numbers from a list.


### Exercise 14
Write a function using `reduce` to find the sum of all numbers in a list.


### Exercise 15
Combine `map`, `filter`, and `reduce` to process a list:
- Square each number.
- Filter numbers greater than 10.
- Compute their sum.
