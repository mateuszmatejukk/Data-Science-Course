## 2. Arithmetic Operators


### Exercise 5
Write a program to calculate the area of a circle based on a radius entered by the user.
Use the formula: `area = pi * r^2`. Ensure `pi = 3.14159`.
Expected Output: "The area of the circle with radius 5 is 78.54."


```python
r = input('Podaj promien: ')
pi = 3.14159
area = pi * float(r) ** 2
formatted_area = format(area, '.2f')
print("The area of the circle with radius for" + ' ' + r + " " + 'is' + " " + formatted_area)
```

### Exercise 6
Given two points (x1, y1) and (x2, y2), calculate the distance between them.
Formula: `distance = sqrt((x2 - x1)^2 + (y2 - y1)^2)`.
Expected Output: "The distance between (1, 1) and (4, 5) is 5.0."


```python
from math import sqrt

x1 = float(input("Podaj x1 :"))
y1 = float(input("Podaj y1 :"))
x2 = float(input("Podaj x2 :"))
y2 = float(input("Podaj y2 :"))
distance = sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)
print(f"The distance between ({x1}, {y1}) and ({x2}, {y2}) is {distance:.2f}.")
```

### Exercise 16
Calculate the BMI (Body Mass Index) and classify it using the following ranges:
- Underweight: < 18.5
- Normal weight: 18.5–24.9
- Overweight: 25–29.9
- Obesity: >= 30
#
Formula: `BMI = weight / height^2`


```python
weight = 90
height = 1.9
BMI = (weight / (height ** 2))
if BMI < 18.5:
    print('Underweight')
elif 18.5 < BMI < 24.9:
    print('Normal weight')
elif 25 < BMI < 29.9:
    print('Overweight')
else:
    print('Obesity')

```
