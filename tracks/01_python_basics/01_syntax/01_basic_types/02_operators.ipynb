XOR logical table in markdow:

| A     | B     | A XOR B |
|-------|-------|---------|
| True  | True  | False   |
| True  | False | True    |
| False | True  | True    |
| False | False | False   |

| A     | B     | A AND B | A OR B | A XOR B | A NAND B |
|-------|-------|---------|--------|---------|----------|
| True  | True  | True    | True   | False   | False    |
| True  | False | False   | True   | True    | True     |
| False | True  | False   | True   | True    | True     |
| False | False | False   | False  | False   | True     |
