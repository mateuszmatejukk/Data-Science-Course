### Exercise 22
Find all 3-letter words in `text = "Tom has a dog and a cat"` using regex.
Expected Output: ["Tom", "has", "dog", "cat"]


```python
import re

text = "Tom has a dog and a cat"
three_letter_words = re.findall(r'\b\w{3}\b', text)

print(three_letter_words)
```

### Exercise 24
Extract the domain from email addresses in the list:

Expected Output: ["example.com", "domain.org"]


```python
emails = ["alice@example.com", "bob@domain.org"]

# Extract domains
domains = [email.split('@')[1] for email in emails]

print(domains)
```

### Exercise 25
Use regex to split `text = "apple;banana|cherry,grape"` on `;`, `|`, or `,`.
Expected Output: ["apple", "banana", "cherry", "grape"]


```python
import re

text = "apple;banana|cherry,grape"

# Split on ;, |, or ,
fruits = re.split(r'[;|,]', text)

print(fruits)
```

### Exercise 26 (Advanced)
Validate phone numbers in the following formats:
- (123) 456-7890
- 123-456-7890
- 123.456.7890
Example valid numbers: "(123) 456-7890", "123-456-7890", "123.456.7890"
phone_numbers = ["(123) 456-7890", "123-456-7890", "123.456.7890", "1234567890"]


```python
import re

phone_numbers = ["(123) 456-7890", "123-456-7890", "123.456.7890", "1234567890"]

# Regex pattern for the allowed formats
pattern = r'^(\(\d{3}\)\s\d{3}-\d{4}|\d{3}-\d{3}-\d{4}|\d{3}\.\d{3}\.\d{4})$'

# Check which numbers are valid
valid_numbers = [num for num in phone_numbers if re.match(pattern, num)]

print(valid_numbers)
```
