## Escape Characters: `\n`, `\t`, `\\`, `\'`, `\"`

```python
print("Hello\nWorld")
print("Hello\tWorld")
print("Hello\\World")
print("Hello\'World")
print("Hello\"World")
```

## String alignment

```python
text = "Align"
text.center(20, "-")  # Centers the string with "-" padding
```

```python
text.ljust(20, "-")  # Left-aligns the string with "-" padding
```

```python
text.rjust(20, "-")  # Right-aligns the string with "-" padding
```

## String testing methods

```python
text = "Hello123"
text.isalnum()  # Check if all characters are alphanumeric
```

```python
text = "12345"
text.isdigit()  # Check if all characters are digits
```

```python
int(text) / 2
```

```python
text / 2
```

```python
text = "hello"
text.islower()  # Check if all characters are lowercase
```

```python
text = "HELLO"
text.isupper()  # Check if all characters are uppercase
```

```python
text = "   "
text.isspace()  # Check if the string contains only whitespace
```
