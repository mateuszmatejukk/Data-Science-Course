# Using itertools

## Product

```python
# "AB" --> # ('A', 'A'), ('A', 'B'), ('B', 'A'), ('B', 'B')
from itertools import product

for pair in product("AB", repeat=3):
    print(pair)
```

## Permutations

```python
import itertools

for perm in itertools.permutations([1, 2, 3]):
    print(perm)
```

## Combinations

```python
import itertools as it

for comb in it.combinations([1, 2, 3], 2):
    print(comb)
print('-' * 100)
for comb in it.combinations_with_replacement([1, 2, 3], 2):
    print(comb)
```

```python
# IS combinations_with_replacement different than product?
# no
```

```python
import itertools
from itertools import product

for pair in product([1,2],repeat = 2):
    print(pair)
```

```python
import itertools as it

for comb in it.combinations_with_replacement([1,2],2):
    print(comb)
```

```python
from itertools import cycle
counter = 0
for c in cycle('ABC'):
    print(c)
    counter += 1
    if counter > 10:
        break
```
