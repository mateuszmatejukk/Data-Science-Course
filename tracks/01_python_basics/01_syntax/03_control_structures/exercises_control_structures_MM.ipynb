### Exercise 1
Write a program to determine if a number is positive, negative, or zero.
Expected Output:
- For `x = 5`: "5 is positive"
- For `x = -3`: "-3 is negative"
- For `x = 0`: "0 is zero"


````python

number = int(input('Enter a number:'))
if number > 0:
    print(f'{number} is posivite')
elif number < 0:
    print(f'{number} is negative')
else:
    print(f'{number} is zero')
````

## 2. Loops


### Exercise 5
Write a `for` loop to calculate the factorial of a number entered by the user.
Example:
- Input: `5`
- Output: `120`


````python
factorial = 1
number = int(input('Enter a number:'))
for i in range(1, number + 1):
    factorial *= i
print(factorial)
````

### Exercise 6
Use a `while` loop to find the first power of 2 greater than a given number.
Example:
- Input: `10`
- Output: `16`


````python
power = 1
number = int(input('Enter a number:'))
while power < number:
    power *= 2
print(power)
````

### Exercise 7
Create a list of squares of numbers from 1 to `n` using both a loop and list comprehension.
Example:
- Input: `n = 5`
- Output: `[1, 4, 9, 16, 25]`


````python
n = int(input('Enter a number:'))
squares = [i ** 2 for i in range(1,n +1)]
print(squares)
````

### Exercise 8
Write a program that iterates through a list of integers and prints:
- The number if it is divisible by 3.
- "Fizz" if it is divisible by 5.
- "FizzBuzz" if divisible by both 3 and 5.


````python
for i in range(1,100):
    if i % 3 == 0 and i % 5 == 0:
        print('FizzBuzz')
    elif i % 3 == 0:
        print('Fizz')
    elif i % 5 == 0:
        print('Buzz')

````

### Exercise 9
Use `else` with a loop to check if a number is prime. If the loop completes without finding a divisor, print "Prime".
Example:

## 4. Advanced Iteration


### Exercise 14
Use `itertools.product` to generate all possible combinations of "AB" and "12".
Expected Output: `('A', '1')`, `('A', '2')`, `('B', '1')`, `('B', '2')`


````python
from itertools import product

for pair in product("AB","12", repeat = 1):
    print(pair)
````

### Exercise 15
Use `itertools.combinations` to select all pairs from a list of numbers.
Example:
- Input: `[1, 2, 3]`
- Output: `(1, 2)`, `(1, 3)`, `(2, 3)`


````python
import itertools as it

for comb in it.combinations([1,2,3],2):
    print(comb)
````
