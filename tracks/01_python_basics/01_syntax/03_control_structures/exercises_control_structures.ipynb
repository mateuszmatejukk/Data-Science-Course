# Control Structures Exercises


## 1. Conditional Statements


### Exercise 1
Write a program to determine if a number is positive, negative, or zero.
Expected Output:
- For `x = 5`: "5 is positive"
- For `x = -3`: "-3 is negative"
- For `x = 0`: "0 is zero"


### Exercise 2
Use a ternary operator to assign "even" or "odd" to a variable based on the value of `x`.
Example:
- For `x = 4`: "x is even"
- For `x = 7`: "x is odd"


### Exercise 3
Implement a match-case statement to classify a grade:
- 90–100: "A"
- 80–89: "B"
- 70–79: "C"
- 60–69: "D"
- Below 60: "F"


### Exercise 4
Use the walrus operator to read a user's input repeatedly and stop only when "quit" is entered.
Example:
```
Enter text: hello
Enter text: quit
```


## 2. Loops


### Exercise 5
Write a `for` loop to calculate the factorial of a number entered by the user.
Example:
- Input: `5`
- Output: `120`


### Exercise 6
Use a `while` loop to find the first power of 2 greater than a given number.
Example:
- Input: `10`
- Output: `16`


### Exercise 7
Create a list of squares of numbers from 1 to `n` using both a loop and list comprehension.
Example:
- Input: `n = 5`
- Output: `[1, 4, 9, 16, 25]`


### Exercise 8
Write a program that iterates through a list of integers and prints:
- The number if it is divisible by 3.
- "Fizz" if it is divisible by 5.
- "FizzBuzz" if divisible by both 3 and 5.


### Exercise 9
Use `else` with a loop to check if a number is prime. If the loop completes without finding a divisor, print "Prime".
Example:
- Input: `11`
- Output: "11 is prime"


## 3. Operators


### Exercise 10
Write a program to check if two numbers are coprime (GCD is 1).
Example:
- Input: `x = 15`, `y = 28`
- Output: "15 and 28 are coprime"


### Exercise 11
Use bitwise operators to check if a number is odd or even.
Example:
- Input: `x = 7`
- Output: "7 is odd"


### Exercise 12
Implement a permissions system using bitwise operators:
- Define constants for READ (0b001), WRITE (0b010), and EXECUTE (0b100).
- Given a permission, check if it includes READ or WRITE access.


### Exercise 13
Demonstrate short-circuit evaluation by:
- Checking if a number is divisible by 2 and greater than 10.
- Ensuring a division operation doesn't raise an error if the divisor is zero.


## 4. Advanced Iteration


### Exercise 14
Use `itertools.product` to generate all possible combinations of "AB" and "12".
Expected Output: `('A', '1')`, `('A', '2')`, `('B', '1')`, `('B', '2')`


### Exercise 15
Use `itertools.combinations` to select all pairs from a list of numbers.
Example:
- Input: `[1, 2, 3]`
- Output: `(1, 2)`, `(1, 3)`, `(2, 3)`


### Exercise 16
Use `itertools.cycle` to repeatedly iterate over the list `["Red", "Green", "Blue"]` and stop after 10 iterations.
