# match statement (python 3.10+)

```python
x = 1

match x:
    case 0:
        print("x is 0")
    case 1:
        print("x is 1")
    case _:
        print("x is not 0 or 1")
# done check if working for strings
```

```python
x = "nazwa"
match x:
    case x if "nazwa" in x:
        print("x contains nazwa")
    case x if "nazwa" not in x:
        print("do not contains nazwa")
```

```python
x = "123"
match x:
    case x if "1" in x:
        print("x contains")
    case x:
        print("x doesn't contain")
```
