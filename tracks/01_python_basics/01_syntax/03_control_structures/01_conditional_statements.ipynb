# match statement (python 3.10+)

```python
x = 1

match x:
    case 0:
        print("x is 0")
    case 1:
        print("x is 1")
    case _:
        print("x is not 0 or 1")
# todo: check if working for strings
```

```python
password = 'hello'
len(password)
if len(password) < 6:
    print(f'password is too short,it has {len(password)} characters')
```

```python
password = 'hello'
n = len(password)
if n < 6:
    print(f'password is too short,it has {n} characters')
```

```python
password = 'hello'

if (n := len(password)) < 6:
    print(f'password is too short,it has {n} characters')
```

```python
diameter = 10
pi = 3.1416
circle = {
    'diameter': diameter,
    'radius': (radius := diameter / 2),
    'area': pi * radius ** 2,
    'circumference': 2 * pi * radius
}
circle
```
