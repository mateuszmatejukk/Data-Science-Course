```python
def repeat_string(s, r):
    if not isinstance(r, int):
        raise TypeError(f"r is not an integer but {type(r)}")
    if r <= 0:
        raise ValueError("r must be positive")
    print(s * r)


repeat_string('x', 3)
# repeat_string('x', -1)
repeat_string('x', 1.5)
```

```python
try:
    repeat_string("hello", "world")
except Exception as e:
    print("programm returned an error")
```
