# Error Handling Exercises


## 1. Basics of Exception Handling


### Exercise 1
Write a program that handles a `ZeroDivisionError` exception.
Prompt the user for two numbers and divide them, ensuring the program doesn't crash if the denominator is zero.


```
try:
    a = float(input("Podaj pierwszą liczbę: "))
    b = float(input("Podaj drugą liczbę: "))

    wynik = a / b
    print("Wynik dzielenia:", wynik)

except ZeroDivisionError:
    print("Błąd: Nie można dzielić przez zero!")

except ValueError:
    print("Błąd: Podano nieprawidłowe dane (to nie jest liczba).")
```

### Exercise 2
Write a program that prompts the user for an integer and handles a `ValueError` if the input is not an integer.


```
try:
    liczba = int(input("Podaj liczbę całkowitą: "))
    print("Wpisałeś liczbę:", liczba)

except ValueError:
    print("Błąd: To nie jest poprawna liczba całkowita!")
```

### Exercise 3
Write a program that tries to open a file named `example.txt` and handles the `FileNotFoundError`.


```
try:
    with open("example.txt", "r") as plik:
        zawartosc = plik.read()
        print("Zawartość pliku:")
        print(zawartosc)

except FileNotFoundError:
    print("Błąd: Plik 'example.txt' nie został znaleziony.")
```

### Exercise 4
Write a program that prompts the user for two numbers and performs addition, subtraction, multiplication, and division.
Handle any exceptions that may occur, such as `ZeroDivisionError` or `ValueError`.


```
try:
    a = float(input("Podaj pierwszą liczbę: "))
    b = float(input("Podaj drugą liczbę: "))

    print("Dodawanie:", a + b)
    print("Odejmowanie:", a - b)
    print("Mnożenie:", a * b)
    print("Dzielenie:", a / b)

except ValueError:
    print("Błąd: Podano nieprawidłowe dane (to nie jest liczba).")

except ZeroDivisionError:
    print("Błąd: Nie można dzielić przez zero!")
```

### Exercise 5
Write a program that handles any arithmetic errors using the `ArithmeticError` exception.


```
try:
    a = float(input("Podaj pierwszą liczbę: "))
    b = float(input("Podaj drugą liczbę: "))

    print("Dodawanie:", a + b)
    print("Odejmowanie:", a - b)
    print("Mnożenie:", a * b)
    print("Dzielenie:", a / b)

except ValueError:
    print("Błąd: Podano nieprawidłowe dane (to nie jest liczba).")

except ArithmeticError:
    print("Błąd arytmetyczny (np. dzielenie przez zero).")
```

### Exercise 6

### Exercise 7
Write a program that demonstrates the use of `try`, `except`, `else`, and `finally` blocks.
Ensure the `finally` block always executes, whether or not an exception occurs.


```
try:
    a = float(input("Podaj pierwszą liczbę: "))
    b = float(input("Podaj drugą liczbę: "))

    wynik = a / b

except ValueError:
    print("Błąd: Podano nieprawidłowe dane.")

except ZeroDivisionError:
    print("Błąd: Nie można dzielić przez zero!")

else:
    print("Wynik dzielenia:", wynik)

finally:
    print("Koniec programu (ten komunikat pojawia się zawsze).")
```
